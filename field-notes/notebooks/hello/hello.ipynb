## A first notebook in Neurodesk

Neurodesk provides a Linux environment for research software. This notebook uses only Python's built-in functionality; there is nothing to install for the example. Heavier analyses can record their Neurodesk tool versions and environment alongside their code.

A notebook kernel runs Python cells and remembers variables between them. A terminal runs shell commands in the Linux environment. Typing Python into the shell is not the same as running a notebook cell.

## Define a function

Run this cell first. It defines `hello`; the function does not print anything until we call it.

```
def hello():
    print("Hello, world!")
```

## Call it

Run the next cell. The expected output is `Hello, world!`. That is an expectation, not a saved execution result on this page.

```
hello()
```

## Working copy and published snapshot

I edit the working copy in native VS Code attached to my local Neurodesk container. Open the repository's `workbench/hello/hello.ipynb`, select a Python kernel inside Neurodesk, and run the cells in order. Restart the kernel and run all cells to check that the notebook does not depend on hidden state.

The workbench is private, ignored scratch space. Publishing explicitly selects files into `content/notes/hello`; the site reads that snapshot, not the live notebook. Outputs, execution counts, attachments, and private notebook metadata are removed. Inspect the selected source for sensitive information before publishing: stripping metadata does not anonymize source code.

The public website is a reader and download surface, not a shared shell. To run the notebook, use your own Neurodesk session and select its Python kernel.
